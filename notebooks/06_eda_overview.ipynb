#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
## Write your code here
df['RemoteWork']=df['RemoteWork'].fillna(df['RemoteWork'].mode()[0])
df['JobSat']=df['JobSat'].fillna(df['JobSat'].median())
missing = df[['Employment', 'JobSat','RemoteWork']].isnull().sum()
missing_pct= (missing / len(df) * 100)
print(missing) 
print(missing_pct)
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
## Write your code here

df['YearsCodePro']= df['YearsCodePro'].replace({"Less than 1 year": 0, "More than 50 years": 51})
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'])

df['ExperienceRange'] = pd.cut(df['YearsCodePro'], bins = [0, 5, 10, 20, 60], labels = ['0-5', '5-10', '10-20', '>20'], include_lowest= True)
df['ExperienceRange'].value_counts()

mean_sat = df.groupby("ExperienceRange")["JobSat"].mean()
mean_sat.plot(kind='bar', color='Blue')
plt.title("Mean Job Satisfaction by Experience")
plt.xlabel("Years of professional Coding")
plt.ylabel("Mean JobSat")
plt.xticks(rotation = 0)
plt.show()

#Changed median to mean because NaN values with .median would be all filled with 7.0: all bars in the chart were the same with no difference.
#With mean you can see a slightly difference over the years.
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
## Write your code here
sns.countplot(data=df, x ="JobSat")
plt.title("Job Satisfaction Distribution")
plt.xlabel("JobSat Score (0-10)")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()

# 7.0 scores come by the NaN filling. True 7.0 scores are 6.500. Real value peak is at 8.0.
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
## Write your code here

countries= ['Germany', 'India', 'Ukraine', 'France']

df_countries =df[ df['Country'].isin(countries)]

df_countries=df_countries.dropna(subset="LanguageHaveWorkedWith")

df_countries['LanguageHaveWorkedWith']= df_countries['LanguageHaveWorkedWith'].str.split(';')

df_lang= df_countries.explode('LanguageHaveWorkedWith', ignore_index= True)

lang_country= pd.crosstab(index = df_lang['LanguageHaveWorkedWith'], columns = df_lang['Country'])

respondents=df_countries['Country'].value_counts()
lang_pct= lang_country / respondents

top_lang= lang_country.sum(axis=1).sort_values(ascending=False).head(10).index

top10_lang=lang_pct.loc[top_lang]


plt.figure(figsize=(8,6))
sns.heatmap(top10_lang, annot=True, fmt='.0%', cmap='Greens')
plt.title("% of Developers Using Each Language by Country")
plt.xlabel("Country") 
plt.xticks(rotation=0)
plt.ylabel("Programming Language") 
plt.show()

#Programming languages by country
# Countries: Germany, India, Ukraine, France (among the top 10 by respondents).
# LanguageHaveWorkedWith is multi-select (";"), so it was split and exploded
# to one row per language. Counts were normalized by the number of respondents
# per country, so values show the % of developers using each language
# (raw counts would favor countries with more respondents, e.g. Germany).
#
# Key findings:
# - JavaScript is the most used language in all four countries (56–67%).
# - Python is strongest in France (57%) and weakest in Ukraine (37%).
# - Bash/Shell splits the countries: ~40% in France/Germany vs ~20% in India/Ukraine.
# - Ukraine shows lower percentages overall, suggesting fewer languages per developer.
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
## Write your code here
# Step 8 - Correlation between experience and job satisfaction
# Keep only rows where both values are present
df_corr = df[['YearsCodePro', 'JobSatPoints_1']].dropna()

# Pearson correlation coefficient (-1 to 1)
corr = df_corr['YearsCodePro'].corr(df_corr['JobSatPoints_1'])
print(f"Correlation between YearsCodePro and JobSatPoints_1: {corr:.3f}")

# Scatter plot with trend line
plt.figure(figsize=(8, 6))
sns.regplot(data=df_corr, x='YearsCodePro', y='JobSatPoints_1', scatter_kws={'alpha': 0.2, 's': 10}, line_kws={'color': 'red'})
plt.title("Years of Professional Coding vs Job Satisfaction Points")
plt.xlabel("Years of Professional Coding")
plt.ylabel("JobSatPoints_1")
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
## Write your code here
# Step 9 - Educational background vs employment type
# Employment is multi-select: keep only the 5 most common categories
top_emp = df['Employment'].value_counts().head(5).index
df_edu = df[df['Employment'].isin(top_emp)].dropna(subset=['EdLevel'])

# Cross-tabulation: % of each education level in each employment type (rows sum to 100%)
edu_emp = pd.crosstab(df_edu['EdLevel'], df_edu['Employment'], normalize='index')

plt.figure(figsize=(12, 6))
sns.heatmap(edu_emp, annot=True, fmt='.0%', cmap='Blues')
plt.title("Employment Type by Education Level")
plt.xlabel("Employment Type")
plt.ylabel("Education Level")
plt.xticks(rotation=30, ha='right')
plt.show()
```
