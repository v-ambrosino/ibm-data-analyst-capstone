### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



**Approach:** multi-select answers are split on `;` and exploded, then the top 5 values are shown as shares of their total.

```python
w_data= df['DatabaseWantToWorkWith'].dropna()
w_data=w_data.str.split(';')
w_data=w_data.explode()
want_data=w_data.value_counts().head(5)
plt.pie(want_data, labels= want_data.index, autopct='%1.1f%%')
plt.title('Database work preference')
plt.axis('equal')
plt.show()
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
df['DevType'].str.contains(';').any()
dev_t=df['DevType'].dropna().value_counts().head(5)
plt.figure(figsize=(12,7))
plt.pie(dev_t, labels=dev_t.index,autopct='%1.1f%%')
plt.title('Type of Developers')
plt.axis('equal')
plt.show()
```

##### 1.3 Create a pie chart for the operating systems used by respondents for professional use


The `OpSysProfessional` use column shows the operating systems developers use professionally. Let’s visualize the distribution of the top operating systems in a pie chart.



```python
op_sys=df['OpSysProfessional use'].dropna()
used_OS=op_sys.str.split(';')
used_OS=used_OS.explode()
used_OS= used_OS.value_counts().head(5)
plt.figure(figsize=(12,7))
plt.pie(used_OS, labels=used_OS.index, autopct='%1.1f%%')
plt.title('Pie for top 5 OS used by professionals')
plt.axis('equal')
plt.show()
```

### Task 2: Additional Visualizations and Comparisons


##### 2.1 Pie Chart for Top 5 Programming Languages Respondents Have Worked With


The `LanguageHaveWorkedWith` column contains the programming languages that respondents have experience with. We’ll plot a pie chart to display the composition of the top 5 languages.



```python
df['LanguageHaveWorkedWith'].str.contains(';').any()
work_lang=df['LanguageHaveWorkedWith'].dropna()
lang_wowith= work_lang.str.split(';')
lang_wowith=lang_wowith.explode()
lang_wowith=lang_wowith.value_counts().head(5)
plt.figure(figsize=(12,7))
plt.pie(lang_wowith, labels=lang_wowith.index, autopct='%1.1f%%')
plt.title('Pie for Top 5 languages used by respondents')
plt.axis('equal') 
plt.show()
```

##### 2.2 Pie Chart for Top Collaboration Tools used in Professional Use

```python
df['NEWCollabToolsHaveWorkedWith'].str.contains(';').any()
new_t=df['NEWCollabToolsHaveWorkedWith'].dropna()
new_tools=new_t.str.split(';')
new_tools=new_tools.explode()
new_tools=new_tools.value_counts().head(5)
plt.figure(figsize=(12,7))
plt.pie(new_tools, labels=new_tools.index, autopct='%1.1f%%')
plt.title('Pie for Top Collab Tools used by respondents', pad= 20)
plt.axis('equal')
plt.show()
```
