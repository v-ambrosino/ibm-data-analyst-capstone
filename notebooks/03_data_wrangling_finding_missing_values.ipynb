### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
df.isnull().sum()
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



**Approach:** a heatmap of the 20 columns with the most missing values (first 1,000 rows) shows where the gaps are concentrated.

```python
import seaborn as sns
import matplotlib.pyplot as plt

top_nan = df.isnull().sum().sort_values(ascending=False).head(20).index

plt.figure(figsize=(14, 6))
sns.heatmap(df[top_nan].head(1000).isnull(), cmap="Blues", cbar=False, yticklabels=False)
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
# Task 4: missing values in the Employment column
df['Employment'].isnull().sum()
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


**Approach:** `Employment` has no missing values, so the mode imputation is shown as a technique and leaves the data unchanged.

```python
# Task 5: most frequent value in Employment
most_frequent = df['Employment'].mode()[0]
print(most_frequent)
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
# Task 6: replace missing values with the most frequent value
df['Employment'] = df['Employment'].fillna(most_frequent)
df['Employment'].isnull().sum()
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
# Task 7: distribution of Employment after imputation (top 10 categories)
plt.figure(figsize=(10, 6))
df['Employment'].value_counts().head(10).sort_values().plot(kind='barh')
plt.title('Employment after Imputation (Top 10)')
plt.xlabel('Number of Respondents')
plt.tight_layout()
plt.show()
```
